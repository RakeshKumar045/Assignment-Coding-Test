```python
# Write a Python class called TextChunker with:
# __init__(self, chunk_size: int, overlap: int)
#
#
# chunk(self, text: str) -> list[str]
#
#
# chunk() should split a text string into chunks of chunk_size characters, where consecutive chunks share overlap characters.


Example: chunk_size=10, overlap=3, text="abcdefghijklmnopqrstu"

Output: ["abcdefghij", "hijklmnopq", "opqrstu"]
```

```python
class TextChunker:
    def __init__(self, chunk_size: int, overlap: int):
        self.chunk_size = chunk_size
        self.overlap = overlap
```

```python
class TextChunker:
  def __init__(self, chunk_size: int, overlap: int):
    self.chunk_size = chunk_size
    self.overlap = overlap

  def chunk(self, text : str) -> list[str]:
    if self.chunk_size <= 0:
      raise ValuseError("chunk size must begreater than")
    if self.overlap >= self.chunk_size :
      raise ValuseError("overlap must be less than chunk size ")

    c = []
    s = 0
    text_len = len(text)

    while s < text_len:
      end  = s + self.chunk_size
      c.append(text[s:end])
      s += self.chunk_size - self.overlap

    return c


input_text="abcdefghijklmnopqrstu"

text_obj = TextChunker(10, 3)
result = text_obj.chunk(input_text)

print(result)







```
