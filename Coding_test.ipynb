```python
l1 = [1, [2, [3, 4], 5], 6]
def flatten(lst):
    res = []
    # print("welcome ..")
    for ind,i in enumerate(lst):
        # print("ind ", ind, " value ", i)
        if isinstance(i,list):
            # print("instance res : ", res, " list ", i)
            res.extend(flatten(i))
            # print("after res : ", res, " list ", i)
        else:
            res.append(i)
            # print("append : ",res)
    print("return res: ", res)
    return res

print(flatten(l1))
```
