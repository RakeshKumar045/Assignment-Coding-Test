```python
def dict_char(s):
  d = {}
  s = s.lower().replace(" ","")
  for i in s:
    if i not in d:
      d[i] = 1
    else:
      d[i] = d[i] + 1
  return d

def check_anagrams(s1, s2):
  d1 = dict_char(s1)
  d2 = dict_char(s2)
  return d1 == d2
```

```python
inp1 = "cat"
inp2 = "tak"
print(check_anagrams(inp1, inp2))
```
