#### 1. Write a python program to count the number of letters in a given string using list only
      Input : 'aabacbdea'
      Expected Output : a:4 b:2 c:1 d:1 e:1

````python
s = 'aabacbdea'
  # Expected Output : a:4 b:2 c:1 d:1 e:1

d = {}

for i in s:
    if i in d:
        d[i] = d[i] + 1
    else:
        d[i] = 1

d
````

### 2. Write a code to reverse a string without using any predefined (available) python functions. Use only lists
        
      Input : 'ABCDCBA'

````python
s = 'ABCDCBA123'
s1 = ""
len_s = len(s)
for i,k in enumerate(s):
    s1 += s[len_s-1-i]

print(s1)
````
