```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.utils import to_categorical
```

```python

sentences = [
    "machine learning is very useful",
    "machine learning is a powerful technology",
    "machine learning is changing the world",
    "deep learning is a subset of machine learning",
    "deep learning uses neural networks",
    "neural networks can learn complex patterns",
    "deep learning models require lots of data",
    "machine learning models learn from data",
    "artificial intelligence is changing technology",
    "artificial intelligence uses machine learning",
]

# Create tokenizer and vocabulary
tokenizer = Tokenizer(oov_token="<OOV>")
tokenizer.fit_on_texts(sentences)

word_index = tokenizer.word_index
vocab_size = len(word_index) + 1

print(word_index)
print("Vocabulary size:", vocab_size)
```

```python
input_sequences = []

for sentence in sentences:
    # Convert sentence to integer tokens
    token_list = tokenizer.texts_to_sequences([sentence])[0]

    # Create examples:
    # [machine] -> learning
    # [machine, learning] -> is
    # etc.
    for i in range(1, len(token_list)):
        input_sequences.append(token_list[:i + 1])

print(input_sequences[:5])
```

```python
# Find the longest sequence length
max_sequence_len = max(len(seq) for seq in input_sequences)

# Pad all sequences to same length
input_sequences = pad_sequences(
    input_sequences,
    maxlen=max_sequence_len,
    padding="pre"
)

# X contains all words except the last word
X = input_sequences[:, :-1]

# y is the last word (the word to predict)
y = input_sequences[:, -1]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nX examples:")
print(X[:5])

print("\ny examples:")
print(y[:5])
```

```python
y = to_categorical(y, num_classes=vocab_size)

print("One-hot y shape:", y.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=32, input_length=max_sequence_len - 1),
    LSTM(64),
    Dense(vocab_size, activation="softmax")
])

model.compile(
    loss="categorical_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)

model.fit(X, y, epochs=200, verbose=1)
```
